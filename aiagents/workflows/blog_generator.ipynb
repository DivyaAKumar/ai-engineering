```python
from langgraph.graph import StateGraph, START, END
from IPython.display import Image
from typing import TypedDict
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
#load environment
load_dotenv("../../LangChainModels/.env")
```

```python
class State(TypedDict):
    topic: str
    prompt: str
    outline : str
    response : str
```

```python
llm = ChatOpenAI()
```

```python
def generate_outline(state: State) -> State:
    topic = state["topic"]
    prompt = f"Generate an outline for the given topic: {topic}"
    outline = llm.invoke(prompt).content
    state["outline"] = outline
    return state
```

```python
def generate_blog(state: State) -> State:
    outline = state["outline"]
    topic = state["topic"]
    prompt = f"Generate a blog post based on the following outline and topic: {outline} Topic: {topic}"
    response = llm.invoke(prompt).content
    state["response"] = response
    return state
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("generate_outline", generate_outline)
```

```python
graph.add_node("generate_blog", generate_blog)
```

```python
graph.add_edge(START, "generate_outline")
graph.add_edge("generate_outline", "generate_blog")
graph.add_edge("generate_blog", END)
```

```python
topic = input("Enter any topic for the blog:")
```

```python
app = graph.compile()
```

```python
result = app.invoke({"topic": topic, "outline": "", "response": ""})
```
