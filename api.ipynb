```python
import requests
def get_temperature(latitude, longitude):
    url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current=temperature_2m"
    response = requests.get(url, timeout=10)
    data = response.json()
    return data["current"]["temperature_2m"]

#get weather for different city entered by the user

input_latitude = float(input("Enter the latitude: "))
input_longitude = float(input("Enter the longitude: "))
temperature = get_temperature(input_latitude, input_longitude)
print(f"The temperature at ({input_latitude}, {input_longitude}) is {temperature}°C.")
```

```python
#finding weather by entering city name 
import requests
city = input("Enter the city name: ")
#to get latitudes and longitudes of the city, we can use th
geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}"
geo_response = requests.get(geo_url, timeout=10)
geo_data = geo_response.json()

latitude= geo_data["results"][0]["latitude"]
longitude= geo_data["results"][0]["longitude"]

def get_temperature(latitude, longitude):
    url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current=temperature_2m"
    response = requests.get(url, timeout=10)
    data = response.json()
    return data["current"]["temperature_2m"]

temperature = get_temperature(latitude, longitude)

print(f"The current temperature in {city} is {temperature}°C.")
```
