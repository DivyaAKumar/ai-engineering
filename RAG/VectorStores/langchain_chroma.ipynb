```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document
from langchain_core.documents import Document

# Create LangChain documents for IPL players

doc1 = Document(
    page_content="Virat Kohli is one of the most successful batsmen in IPL history. He has played for Royal Challengers Bengaluru and is known for his aggressive batting style and consistency.",
    metadata={"team": "Royal Challengers Bengaluru"}
)

doc2 = Document(
    page_content="Rohit Sharma is one of the most successful captains in IPL history, leading Mumbai Indians to five IPL titles. He is also known for his elegant batting style.",
    metadata={"team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. He is known for his finishing skills and excellent wicketkeeping.",
    metadata={"team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his accuracy, yorkers, and variations.",
    metadata={"team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, he is also an excellent fielder.",
    metadata={"team": "Chennai Super Kings"}
)

doc6 = Document(
    page_content="Shubman Gill is a talented Indian batsman known for his elegant stroke play and consistency. He has represented Gujarat Titans in the IPL.",
    metadata={"team": "Gujarat Titans"}
)

doc7 = Document(
    page_content="Hardik Pandya is a powerful all-rounder known for his explosive batting and effective bowling. He has represented Gujarat Titans and Mumbai Indians in the IPL.",
    metadata={"team": "Mumbai Indians"}
)

doc8 = Document(
    page_content="Rishabh Pant is an aggressive left-handed batsman and wicketkeeper. He is known for his attacking batting and has represented Delhi Capitals in the IPL.",
    metadata={"team": "Delhi Capitals"}
)

doc9 = Document(
    page_content="KL Rahul is a stylish top-order batsman and wicketkeeper. He is known for his ability to score quickly and has represented Lucknow Super Giants in the IPL.",
    metadata={"team": "Lucknow Super Giants"}
)

doc10 = Document(
    page_content="Sanju Samson is a talented wicketkeeper-batsman and captain. He is known for his powerful batting and has represented Rajasthan Royals in the IPL.",
    metadata={"team": "Rajasthan Royals"}
)
```

```python
from google.colab import userdata

OPENAI_API_KEY = userdata.get("OPENAI_API_KEY")
```

```python
docs = [doc1,doc2,doc3,doc4,doc5,doc6,doc7,doc8,doc9,doc10]
```

```python
embeddings = OpenAIEmbeddings(
    api_key=OPENAI_API_KEY
)

vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="my_chroma_db",
    collection_name="players"
)
```

```python
#add documents
vector_store.add_documents(docs)
```

```python
#view documents
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
#search documents
vector_store.similarity_search(query= "Who is Virat Kohli?", k=1) #if k=1 only one similar document will be shown if k=2 then 2 documents are shown
```

```python
#search with a similarity score
vector_store.similarity_search_with_score(query='who among these are bowler?', k=2) #similarity score jitna kam utna acha this means 0.3 is better than 0.4
```

```python
#meta data filtering
vector_store.similarity_search_with_score(
    query = "",
    filter = {
        'team': 'Mumbai Indians'
    }
)
```

```python
#update documents
updated_doc1 = Document(
    page_content="Virat Kohli is handsome",
    metadata={"team": "Mumbai Indians"}
)
vector_store.update_document(document_id='b676dd2f-d3f3-4b38-9e11-6029a6447bae', document = updated_doc1)
```

```python
#view a particular document
result = vector_store.get(
    include=["documents", "metadatas"]
)

print(result["documents"][0])
print(result["metadatas"][0])
```

```python
#delete document
vector_store.delete(ids=['b676dd2f-d3f3-4b38-9e11-6029a6447bae'])
```
