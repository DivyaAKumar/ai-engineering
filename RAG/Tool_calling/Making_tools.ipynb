Custom Tools

```python
from langchain_core.tools import tool
```

```python
#create a function
def multiply(a,b):
  """multiply two numbers""" #highly recommended to add doc string
  return a*b
```

```python
#add type hints
  """multiply two numbers"""
def multiply(a:int, b:int) -> int:
  return a*b
```

```python
#add tool decorator
@tool
def multiply(a:int, b:int) -> int:
    """multiply two numbers"""
    return a*b
```

```python
result = multiply.invoke({'a': 3, 'b': 4})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

Using StructuredTool Class

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
  a: int = Field(required = True, description='first number')
  b: int = Field(required = True,description='second number')
```

```python
def multiply_num(a:int, b:int) -> int:
  """multiply two numbers"""
  return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_num,
    name = 'multiply',
    description = 'multiply two numbers',
    args_schema = MultiplyInput
)
```

```python
result = multiply_tool.invoke({'a': 3, 'b': 4})
print(result)
```

Using BaseTool Class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
#args schema using pydantic
class MultiplyInput(BaseModel):
  a: int = Field(required = True, description='first number')
  b: int = Field(required = True,description='second number')
```

```python
class MultiplyTool(BaseTool):
  name : str = 'multiply'
  description : str = 'multiply two numbers'
  args_schema: Type[BaseModel] = MultiplyInput

  def _run(self, a:int, b:int) -> int:
    """multiply two numbers"""
    return a*b
```

```python
multiply_tool = MultiplyTool()
```

```python
result_basetool = multiply_tool.invoke({'a': 3, 'b': 4})
print(result_basetool)
```

ToolKit

```python
from langchain_core.tools import tool

#custom tools
@tool
def add(a:int, b:int) -> int:
  """add two numbers"""
  a+b

@tool
def multiply(a:int, b:int) -> int:
  """multiply two numbers"""
  a*b
```

```python
class MathToolkit:
  def get_tools(self):
    return[add,multiply]
```

```python
toolkit = MathToolkit()
```

```python
tools = toolkit.get_tools()
```

```python
for tool in tools:
  print(tool.name, '=>', tool.description)
```
